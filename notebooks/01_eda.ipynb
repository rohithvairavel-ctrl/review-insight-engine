# 01 — Exploratory Data Analysis

Explore product review text, score distribution, and sentiment labels.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT))
import pandas as pd
from src.load import load_reviews
from src.clean import clean_text, tokenize

df = load_reviews(prefer_full=True)
df.head()
```

```python
print('Rows:', len(df))
print(df['sentiment'].value_counts())
print(df['score'].describe())
df['text_len'] = df['text'].str.len()
df['token_count'] = df['text'].map(lambda t: len(tokenize(t)))
df[['text_len','token_count']].describe()
```

```python
import matplotlib.pyplot as plt
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
df['sentiment'].value_counts().reindex(['negative','neutral','positive']).plot(kind='bar', ax=axes[0], color=['#E45756','#F58518','#54A24B'])
axes[0].set_title('Sentiment counts')
df['score'].value_counts().sort_index().plot(kind='bar', ax=axes[1], color='#4C78A8')
axes[1].set_title('Star ratings')
plt.tight_layout()
plt.show()
```
